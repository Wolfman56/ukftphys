```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import constants

# Physical Constants
c = constants.c          # Speed of light (m/s)
h = constants.h          # Planck constant (J s)
hbar = constants.hbar    # Reduced Planck constant (J s)
G = constants.G          # Gravitational constant (m^3 kg^-1 s^-2)
kB = constants.k         # Boltzmann constant (J/K)
eV = constants.e         # Electron volt (J)
GeV = 1e9 * eV           # Giga-electron volt (J)
M_sun = 1.989e30         # Solar mass (kg)
M_pl = np.sqrt(hbar * c / G) # Planck Mass (kg)

print(f"Planck Mass: {M_pl:.4e} kg")
```

## 1. The 30 GeV Black Hole
The Hawking Temperature of a Schwarzschild Black Hole is given by:
$$ T_H = \frac{\hbar c^3}{8 \pi G M k_B} $$

We want to find the mass $M_{BH}$ such that $k_B T_H \approx 30$ GeV.
Rearranging for $M$:
$$ M = \frac{\hbar c^3}{8 \pi G (k_B T_H)} $$

```python
def hawking_temp_to_mass(T_GeV):
    # Convert T from GeV to Joules (E = kB T)
    E_thermal = T_GeV * GeV
    
    # M = (hbar c^3) / (8 pi G E_thermal)
    numerator = hbar * c**3
    denominator = 8 * np.pi * G * E_thermal
    
    return numerator / denominator

target_temp_GeV = 30.0
mass_30GeV_BH = hawking_temp_to_mass(target_temp_GeV)

print(f"Target Temperature: {target_temp_GeV} GeV")
print(f"Corresponding Black Hole Mass: {mass_30GeV_BH:.4e} kg")
print(f"In Planck Masses: {mass_30GeV_BH / M_pl:.4f} M_pl")
print(f"In Grams: {mass_30GeV_BH * 1000:.4f} g")
```

### Interpretation
A Black Hole radiating at 30 GeV (the scale of our Monopole) would have a mass of roughly $3.5 \times 10^{11}$ kg (a small mountain/asteroid). However, checking the calculation carefully:

The formula usually yields huge temperatures for tiny masses.
Let's check: $T \propto 1/M$.
So a *higher* temperature implies a *smaller* mass.
If $T \approx 30$ GeV, the mass must be very small. Let's verify the output.

```python
def mass_to_hawking_temp(M_kg):
    return (hbar * c**3) / (8 * np.pi * G * M_kg * kB)

# Check Solar Mass BH Temp
T_sun = mass_to_hawking_temp(M_sun)
print(f"Solar Mass BH Temp: {T_sun:.4e} K ({T_sun * kB / eV:.4e} eV)")

# Check Planck Mass BH Temp
T_pl = mass_to_hawking_temp(M_pl)
print(f"Planck Mass BH Temp: {T_pl:.4e} K ({T_pl * kB / GeV:.4e} GeV)")

# Re-calculate 30 GeV Mass
M_target = hawking_temp_to_mass(30.0)
print(f"Mass for 30 GeV Temp: {M_target:.4e} kg")
```

## 2. Evaporation Time & Missing Energy
The lifetime of a Black Hole is given by:
$$ t_{evap} \approx \frac{5120 \pi G^2 M_0^3}{\hbar c^4} $$

If our monopole *is* a tiny primordial black hole (PBH) stable at this mass due to topological constraints (relic), then it wouldn't evaporate. But if it *decays* (as a particle), we can compare its lifetime.

In Exp 48, we calculated the Monopole lifetime decay $\tau \approx 10^{-22}$ s.
Let's see the evaporation time of a BH with this mass.

```python
def evaporation_time(M_kg):
    return (5120 * np.pi * G**2 * M_kg**3) / (hbar * c**4)

t_evap_30GeV = evaporation_time(M_target)
print(f"Evaporation Time for 30 GeV BH: {t_evap_30GeV:.4e} s")
```

### Comparison
There is a huge discrepancy. The lifetime of a BH with $T_H=30$ GeV is extremely long compared to the particle decay time ($10^{-22}$s).
Unless... the Monopole is *not* the Black Hole itself, but the *quantum of radiation* emitted by a larger geometry.

But let's look at the **Missing Energy** hypothesis.
If the Monopole decays largely into the "Mirror Sector" (Invisible), this mimics Hawking Radiation (information entering the BH or thermal radiation escaping to infinity), where the energy is "lost" to our detectors.

Let's plot the Blackbody Spectrum for $T = 30$ GeV to see what the energy distribution of emitted particles would look like.

```python
def planck_spectrum(E_GeV, T_GeV):
    # Normalized Planck distribution I(E) ~ E^3 / (exp(E/T) - 1)
    # Note: For massless particles in 3D
    return (E_GeV**3) / (np.exp(E_GeV / T_GeV) - 1)

energies = np.linspace(0.1, 200, 500) # GeV
spectrum = planck_spectrum(energies, 30.0)

plt.figure(figsize=(10, 6))
plt.plot(energies, spectrum, label='Hawking Radiation (T=30 GeV)')
plt.axvline(x=30.0, color='r', linestyle='--', label='Peak Temperature')
plt.title('Theoretical Emission Spectrum for T_H = 30 GeV')
plt.xlabel('Particle Energy (GeV)')
plt.ylabel('Intensity (Arbitrary Units)')
plt.legend()
plt.grid(True)
plt.show()
```
