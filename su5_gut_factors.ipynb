# SU(5) GUT Normalization and Factors Analysis

This notebook investigates the group theoretic factors in SU(5) Grand Unified Theories, specifically verifying the normalization of Hypercharge and the standard theoretical prediction for the weak mixing angle. We will look for the emergence of factors like 5/9 or 9/5.

```python
# Define Standard Model Particle Charges (Generation 1, Left-Handed States)
# For calculations we use the standard conventions and represent all fields as Left-Handed Weyl Spinors.
# The particles are:
# - Quark doublet Q_L = (u_L, d_L): (3 colors, SU(2) doublet)
# - Up-type singlet u^c_L = anti(u_R): (3 colors, SU(2) singlet)
# - Down-type singlet d^c_L = anti(d_R): (3 colors, SU(2) singlet)
# - Lepton doublet L_L = (nu_L, e_L): (1 color, SU(2) doublet)
# - Electron singlet e^c_L = anti(e_R): (1 color, SU(2) singlet)

particles = []

# Q_L (3 colors)
# u_L: T3 = +1/2, Q = +2/3
particles.extend([{'name': 'u_L', 'T3': 0.5, 'Q': 2/3}] * 3)
# d_L: T3 = -1/2, Q = -1/3
particles.extend([{'name': 'd_L', 'T3': -0.5, 'Q': -1/3}] * 3)

# u^c_L (anti-u_R) (3 colors)
# Charge of anti-particle is opposite to particle. Q(u_R) = 2/3 -> Q(u^c) = -2/3
# T3 for singlet is 0
particles.extend([{'name': 'u_c_L', 'T3': 0.0, 'Q': -2/3}] * 3)

# d^c_L (anti-d_R) (3 colors)
# Q(d_R) = -1/3 -> Q(d^c) = +1/3
particles.extend([{'name': 'd_c_L', 'T3': 0.0, 'Q': 1/3}] * 3)

# L_L (1 color)
# nu_L: T3 = +1/2, Q = 0
particles.append({'name': 'nu_L', 'T3': 0.5, 'Q': 0.0})
# e_L: T3 = -1/2, Q = -1
particles.append({'name': 'e_L', 'T3': -0.5, 'Q': -1.0})

# e^c_L (anti-e_R) (1 color)
# Q(e_R) = -1 -> Q(e^c) = +1
particles.append({'name': 'e_c_L', 'T3': 0.0, 'Q': 1.0})

# Total number of particles should be 15
print(f"Total number of particles: {len(particles)}")
for p in particles:
    print(p)
```

```python
# Section: Calculate Sum of Squared Electric Charges and Weak Isospin

sum_Q2 = sum(p['Q']**2 for p in particles)
sum_T3_sq = sum(p['T3']**2 for p in particles)

print(f"Sum of Q^2: {sum_Q2}")
print(f"Sum of T3^2: {sum_T3_sq}")

# Verify user calculation
# User had: 16/3 = 5.333...
user_calc = 16/3
print(f"User calc: {user_calc}")
print(f"Matches? {abs(sum_Q2 - user_calc) < 1e-10}")

# Calculate Hypercharge Y = Q - T3
# Note: Different conventions exist for Y (factor of 2 usually).
# Here we use Q = T3 + Y_SM.
for p in particles:
    p['Y_SM'] = p['Q'] - p['T3']

sum_Y_SM_sq = sum(p['Y_SM']**2 for p in particles)
print(f"Sum of Y_SM^2: {sum_Y_SM_sq}")
```

```python
# Section: SU(5) Normalization Factor Calculation

# Standard Model generators T_a must be normalized to same value in GUT.
# Tr(T_a T_b) = C delta_ab
# Usually C = 1/2 for fundamental representation of SU(N).
# But here we sum over the fermion representation.

# We require Tr(T_3^2) = Tr( (c * Y_SM)^2 ) = c^2 * Tr(Y_SM^2)
# c is the normalization constant for the U(1) generator to fit into the SU(5) group.
# Often denoted as sqrt(3/5) or similar.

print(f"Tr(T3^2) = {sum_T3_sq}")
print(f"Tr(Y_SM^2) = {sum_Y_SM_sq}")

k_factor = sum_T3_sq / sum_Y_SM_sq
print(f"Normalization factor k = Tr(T3^2) / Tr(Y^2) = {k_factor}")

# The user mentioned standard factors like 3/8, 5/3 etc.
# Check fraction representation
from fractions import Fraction
print(f"k as fraction: {Fraction(k_factor).limit_denominator()}")

# Relation to Weak Mixing Angle
# sin^2(theta_W) = e^2 / g^2 = g'^2 / (g^2 + g'^2) ?
# Actually sin^2(theta_w) = Tr(T3^2) / Tr(Q^2) at GUT scale.
sin2_theta_w_gut = sum_T3_sq / sum_Q2
print(f"Calculated sin^2(theta_W) at GUT scale: {sin2_theta_w_gut}")
print(f"Fraction: {Fraction(sin2_theta_w_gut).limit_denominator()}")
```
